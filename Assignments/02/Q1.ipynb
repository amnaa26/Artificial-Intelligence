```python
def evaluate(x):
    return -1 * (x - 7) ** 2 + 49  
    
def peak_search(n):
    left = 0
    right = n
    while left < right:
        mid = (left + right) // 2
        val = evaluate(mid)
        
        if val >= evaluate(mid + 1) and val >= evaluate(mid - 1):
            return mid
        elif val > evaluate(mid + 1):
            right = mid
        else:
            left = mid + 1

    return left

index = peak_search(50)
print(f"Peak found at index {index}")
```
